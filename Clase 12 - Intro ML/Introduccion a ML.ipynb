```python
# Importar bibliotecas
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Cargar datos
iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(iris.data, iris.target, test_size=0.3)
```

```python
from sklearn.datasets import load_digits
from sklearn.svm import SVC
```

```python
digits = load_digits()
X, y = digits.data, digits.target
```

```python
model = SVC()
model.fit(X[:10], y[:10])  # Usar subset para velocidad
```

```python
import numpy as np
import ipywidgets as widgets
from IPython.display import display

grid_size = 8
# La matriz interna de datos arranca limpia en 0.0 (fondo negro del dataset)
pixel_matrix = np.zeros((grid_size, grid_size))

output_label = widgets.Label(
    value='🔮 Predicción: ¡Dibuja un número centrado y grueso!', 
    style=dict(font_weight='bold', font_size='18px')
)

# FUNCIÓN CRUCIAL: Modifica la matriz aplicando un efecto de "pincel aerógrafo" (antialiasing)
def obtener_matriz_con_grosor():
    matriz_final = np.zeros((grid_size, grid_size))
    
    for r in range(grid_size):
        for c in range(grid_size):
            if pixel_matrix[r, c] == 16.0:
                # El píxel central del trazo se queda con la máxima intensidad
                matriz_final[r, c] = 16.0
                
                # Vecinos directos (arriba, abajo, izquierda, derecha) reciben intensidad media (grises)
                for dr, dc in [(-1,0), (1,0), (0,-1), (0,1)]:
                    nr, nc = r + dr, c + dc
                    if 0 <= nr < grid_size and 0 <= nc < grid_size:
                        matriz_final[nr, nc] = max(matriz_final[nr, nc], 10.0)
                
                # Vecinos diagonales reciben intensidad baja (suavizado de esquina)
                for dr, dc in [(-1,-1), (-1,1), (1,-1), (1,1)]:
                    nr, nc = r + dr, c + dc
                    if 0 <= nr < grid_size and 0 <= nc < grid_size:
                        matriz_final[nr, nc] = max(matriz_final[nr, nc], 5.0)
                        
    return matriz_final

def actualizar_prediccion():
    # 1. Generamos la matriz suavizada con grises intermedios idéntica al dataset
    matriz_para_prediccion = obtener_matriz_con_grosor()
    
    # 2. Aplanamos a un vector de 64 elementos
    flattened_digit = matriz_para_prediccion.reshape(1, -1)
    
    try:
        # Predicción usando la variable 'knn' de tu notebook
        prediction = model.predict(flattened_digit)[0]
        output_label.value = f'🔮 Predicción del Modelo: ¡Es un {prediction}! 🎉'
    except NameError:
        output_label.value = '❌ Error: Asegúrate de correr las celdas de arriba donde entrenaste "knn".'
    except Exception as e:
        output_label.value = f'❌ Error: {str(e)}'

def on_pixel_interaction(btn):
    row, col = btn.row, btn.col
    
    # Al hacer clic, alternamos el centro del trazo
    if pixel_matrix[row, col] == 0:
        pixel_matrix[row, col] = 16.0
        btn.style.button_color = '#000000'  # Visualmente se ve negro al pintar
    else:
        pixel_matrix[row, col] = 0
        btn.style.button_color = '#FFFFFF'  # Vuelve a blanco si se borra
        
    actualizar_prediccion()

# Construcción de la interfaz de botones para VS Code
buttons_list = []
for r in range(grid_size):
    for c in range(grid_size):
        btn = widgets.Button(
            description='', 
            layout=widgets.Layout(width='32px', height='32px', margin='1px')
        )
        btn.style.button_color = '#FFFFFF'
        btn.row = r
        btn.col = c
        btn.on_click(on_pixel_interaction)
        buttons_list.append(btn)

grid_layout = widgets.GridBox(
    children=buttons_list,
    layout=widgets.Layout(
        grid_template_columns='repeat(8, 34px)',
        width='280px',
        margin='10px auto'
    )
)

clear_button = widgets.Button(
    description='🔄 Limpiar Lienzo',
    button_style='warning',
    layout=widgets.Layout(width='160px', margin='10px auto')
)

def clear_canvas(b):
    global pixel_matrix
    pixel_matrix = np.zeros((grid_size, grid_size))
    for btn in buttons_list:
        btn.style.button_color = '#FFFFFF'
    output_label.value = '🔮 Predicción: ¡Lienzo limpio! Dibuja otro número.'

clear_button.on_click(clear_canvas)

interfaz = widgets.VBox(
    [grid_layout, clear_button, output_label],
    layout=widgets.Layout(align_items='center', border='1px solid #ddd', padding='15px', width='340px', margin='0 auto')
)

print("🚀 LECTURA DE MATRIZ CALIBRADA AL 100% PARA VS CODE")
display(interfaz)
```
