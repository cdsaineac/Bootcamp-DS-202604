# Introducción a la Visualización de Datos

## ¿Por qué visualizar datos?

- Detectar patrones
- Comunicar hallazgos
- Tomar decisiones
- Identificar errores en datos

---


# Principios de Buenas Visualizaciones

## 1. Claridad > Estética
- Evitar gráficos recargados
- Menos es más

## 2. Elegir el gráfico correcto
- Comparación de variables categoricas → Gráfico de barras
- Distribución de una variable númerica continua → histogramas
- Relación de variables númericas → Gráfico de dispersión / scatterplot
- Composición → Gráfica de torta (con cuidado)
- Detectar valores atípicos de una variable númerica → Gráfico de Cajas y Bigotes

## 3. Contexto **siempre**
- Títulos claros
- Etiquetas en ejes
- Leyendas

## 4. No distorsionar datos
- No manipular escalas
- Evitar gráficos engañosos

## 5. Consistencia visual
- Colores coherentes
- Tipografías uniformes

# Hora de Practicar

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
# Cargar dataset de ejemplo
df = sns.load_dataset("tips")

df.head()
```

```python
# Ejemplo básico: Histograma

plt.figure(figsize=(8,5))
plt.hist(df['total_bill'], bins=20)

#plt.title("Distribución del total de la cuenta")
#plt.xlabel("Total Bill")
#plt.ylabel("Frecuencia")

plt.show()
```

# Errores comunes

- Demasiados colores
- No etiquetar ejes
- Gráficos innecesarios
- Escalas truncadas
- Uso incorrecto de pie charts

---

## Ejemplo:
Un gráfico sin título ni contexto pierde valor completamente.

```python
# Mejor versión con Seaborn

plt.figure(figsize=(8,5))
sns.histplot(df['total_bill'], kde=True)

plt.title("Distribución del Total de la Cuenta")
plt.xlabel("Valor ($)")
plt.ylabel("Cantidad de casos")

plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x='day', y='total_bill', data=df)

plt.title("Promedio de cuenta por día")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='total_bill', y='tip', hue='sex', data=df)

plt.title("Relación entre cuenta total y propina")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='day', y='total_bill', data=df)

plt.title("Distribución por día")
plt.show()
```

# Uso del Color

## Buenas prácticas:

- Usar colores con propósito
- No usar más de 3-5 colores
- Usar contraste para resaltar
- Evitar colores innecesarios

---

## En Power BI / Excel:
Esto se traduce en:
- Paletas consistentes
- Uso de color para destacar insights

```python
# Ejemplo de uso de color intencional

plt.figure(figsize=(8,5))
sns.barplot(x='day', y='total_bill', data=df, palette='Blues')

plt.title("Uso de paleta monocromática")
plt.show()
```
